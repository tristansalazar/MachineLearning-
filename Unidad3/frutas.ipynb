```python
## Fruit dataset
- Es uno de los conjuntos de datos más conocidos en el mundo del aprendizaje automático y la ciencia de datos.
- Consta de 59 registros con 4 tipos de frutas.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

```python
frutas = pd.read_csv("fruit_data_with_colors.csv")
frutas.head()
```

```python
## Modelo de clasificación usando parámetro k = 1, es decir, 1 vecino
### Selección de variables predictoras (x) y variable a predecir (y)
```

```python
X = frutas[["height", "width", "mass", "color_score"]]
y = frutas["fruit_name"]
```

```python
### Dividir dataset en conjunto de entrenamiento y prueba
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
### Entrenamiento del modelo
- Entrenar un modelo se refiere al proceso de ajustar un modelo a los datos, de modo que el modelo pueda aprender a hacer predicciones basadas en los datos.
- En el aprendizaje automático, esto normalmente se hace optimizando los parámetros del modelo para minimizar el error entre las predicciones del modelo y los valores reales.
```

```python
clasificador_KNN = KNeighborsClassifier(n_neighbors= 1)

clasificador_KNN.fit(X_train, y_train)
print("Precision del clasificador en fase de entrenamiento",
      clasificador_KNN.score(X_train, y_train) )
```

```python
### Prueba del Modelo
La prueba del modelo implica evaluar su desempeño en un conjunto de datos no visto previamente (el conjunto de prueba) para medir qué tan bien generaliza al trabajar con datos nuevos.

**Generalización:** La capacidad del modelo para hacer predicciones precisas en datos nuevos. Un modelo bien probado no solo memoriza los datos de entrenamiento, sino que entiende los patrones subyacentes.
- **Sobreajuste (Overfitting):** El modelo aprende demasiado bien los detalles y ruido de los datos de entrenamiento, lo que lleva a un mal desempeño en el conjunto de prueba.
- **Subajuste (Underfitting):** El modelo no captura suficientemente la relación entre los datos, llevando a un bajo desempeño en ambos conjuntos.
```

```python
y_pred = clasificador_KNN.predict(X_test)

print("Reporte de métricas del clasificador: \n", classification_report(y_test, y_pred))

print( "Precisión:", clasificador_KNN.score(X_test, y_test) )
```

```python
### Reporte de los resultados

**Qué significa cada métrica:**
- **Precision:** De todas las veces que el modelo predijo una fruta, ¿cuántas veces acertó?
- **Recall:** De todas las frutas reales de una clase, ¿cuántas fueron correctamente detectadas?
- **F1-score:** Media armónica entre precisión y recall.
- **Support:** Número de ejemplos de cada clase en el conjunto de prueba.

**Accuracy = 0.53:** de 15 ejemplos de prueba, el modelo acertó 8.

- **apple:** Precision 0.00, Recall 0.00, F1 0.00 (4 muestras). El modelo no identificó ninguna manzana.
- **lemon:** Precision 1.00, Recall 0.60, F1 0.75 (5 muestras). Todo lo que predijo como limón era limón, pero solo detectó el 60%.
- **mandarin:** Precision 1.00, Recall 1.00, F1 1.00 (2 muestras). Desempeño perfecto, aunque con pocas muestras.
- **orange:** Precision 0.38, Recall 0.75, F1 0.50 (4 muestras). Detectó la mayoría de las naranjas, pero muchas predicciones de "orange" eran otras frutas.

**Conclusiones:**
- El modelo no acierta ninguna manzana.
- Lemon y mandarin están bien identificadas.
- Apple y orange se confunden entre sí.
```

```python
### Matriz de confusión
Las filas representan las clases reales y las columnas las predicciones:
- **Diagonal principal:** predicciones correctas (True Positives).
- **Fuera de la diagonal:** predicciones incorrectas (falsos positivos y falsos negativos).
```

```python
print ("Matriz de confusión:\n", confusion_matrix(y_test, y_pred), "\n")
```

```python
### Interpretación

| Fruta real | Correctas | Errores |
|---|---|---|
| apple | 0/4 | Las 4 clasificadas como orange |
| lemon | 3/5 | 1 como apple, 1 como orange |
| mandarin | 2/2 | Ninguno |
| orange | 3/4 | 1 como apple |

El modelo confunde manzanas y naranjas, probablemente porque tienen tamaño y masa similares.

### Calcular la importancia de características (Permutation Importance)
Este método mide cuánto baja la precisión del modelo si se altera aleatoriamente una variable. Cuanto mayor sea esa caída, más importante es la variable.
```

```python
from sklearn.inspection import permutation_importance

result = permutation_importance(clasificador_KNN, X_test, y_test, n_repeats=30, random_state=42)

importancia = pd.DataFrame({
    "Variable": X.columns,
    "Importancia Media": result.importances_mean,
    "Desviación": result.importances_std
}).sort_values("Importancia Media", ascending=False)

print(importancia)
```

```python
import plotly.express as px
fig4 = px.bar(importancia, x="Variable", y="Importancia Media",
              error_y="Desviación", title="Importancia de características (KNN - Permutation Importance)",
              text_auto=".3f", color="Variable")
fig4.update_layout(width=600, height=400)
fig4.show()
```

```python
for k in range(1, 20):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    print(f"k={k} → Precisión: {knn.score(X_test, y_test):.2f}")
```

```python
### Interpretación de estos resultados
- **k muy pequeño (1–2):** El modelo se sobreajusta, clasifica casi idéntico a los ejemplos de entrenamiento.
- **k intermedio (3–5):** El modelo alcanza su máxima estabilidad y generalización, con la precisión más alta (0.60).
- **k grande (>6):** El modelo empieza a subajustar, promedia demasiados vecinos y pierde la capacidad de distinguir clases con pocos datos. Precisión baja (0.40–0.47).

## Otras alternativas para intentar mejorar el rendimiento
- Probar distintas métricas de distancia. Por defecto KNN usa distancia Euclidiana, pero se puede cambiar, por ejemplo: metric='manhattan'.
- Realizar la normalización de datos.
```

```python
from sklearn.preprocessing import StandardScaler

X = frutas[["height", "width", "mass", "color_score"]]
y = frutas["fruit_name"]

X_train, X_test, y_train, y_test = train_test_split( X, y, random_state=42,
                                                    stratify=y)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

knn_scaled = KNeighborsClassifier(n_neighbors=3)
knn_scaled.fit(X_train_scaled, y_train)
y_pred_scaled = knn_scaled.predict(X_test_scaled)

print("\nReporte de clasificación:")
print(classification_report(y_test, y_pred_scaled))
```

```python
### Interpretación
El modelo KNN con datos escalados funciona mucho mejor. KNN mide distancias entre puntos, por lo que si las variables no están en la misma escala, las que tienen valores más grandes (como mass) dominan la distancia total.

| Clase | Precision | Recall | F1-score | Interpretación |
|---|---|---|---|---|
| apple | 1.00 | 0.80 | 0.89 | Todo lo que predijo como apple era apple, pero se le escapó una manzana |
| lemon | 1.00 | 1.00 | 1.00 | Perfecto |
| mandarin | 1.00 | 1.00 | 1.00 | Perfecto, aunque solo hay 1 muestra |
| orange | 0.83 | 1.00 | 0.91 | Detectó todas las naranjas, pero una manzana la clasificó como naranja |
```

```python
print ("Matriz de confusión:\n", confusion_matrix(y_test, y_pred_scaled), "\n")
```

```python
### Interpretación

| Clase real | Correctas | Incorrectas |
|---|---|---|
| apple | 4/5 | 1 clasificada como orange |
| lemon | 4/4 | 0 |
| mandarin | 1/1 | 0 |
| orange | 5/5 | 0 |

**Conclusiones:**
- El modelo identifica correctamente el 93% de las frutas.
- Todas las clases están bien representadas gracias a stratify=y.
- El escalado fue clave para mejorar el rendimiento.
- Solo hay una confusión: una manzana que se parece demasiado a una naranja en altura, ancho o color_score.

## Importancia de las variables después del escalado
```

```python
from sklearn.inspection import permutation_importance

result = permutation_importance(knn_scaled, X_test_scaled, y_test, n_repeats=30, random_state=42)

importancia = pd.DataFrame({
    "Variable": X.columns,
    "Importancia Media": result.importances_mean,
    "Desviación": result.importances_std
}).sort_values("Importancia Media", ascending=False)

print(importancia)
```

```python
import plotly.express as px
fig4 = px.bar(importancia, x="Variable", y="Importancia Media",
              error_y="Desviación", title="Importancia de características (KNN - Permutation Importance)",
              text_auto=".3f", color="Variable")
fig4.update_layout(width=600, height=400)
fig4.show()
```

```python
### Interpretación
- **Antes del escalado**, la variable mass dominaba el modelo porque sus valores son mucho mayores (100–300) frente a color_score (0–1).
- **Después del escalado**, todas las variables están en la misma escala, y la importancia se distribuye entre las demás.

| Variable | Antes de escalar | Después de escalar | Interpretación |
|---|---|---|---|
| mass | 0.267 | 0.038 | Pierde peso: antes dominaba solo por su rango numérico |
| height | -0.029 | 0.262 | Se vuelve la más importante |
| width | -0.062 | 0.189 | Gana importancia |
| color_score | -0.004 | 0.182 | Gana importancia |

Tras escalar, height, width y color_score aportan información útil. Este ajuste permitió que el modelo KNN mejorara su precisión de 60% a 93%.

## Implementación del modelo para Predicciones sobre nuevas frutas (ficticios)
```
